## 度数分布表
度数分布の作成手順
- 範囲を求める：範囲=最大値-最小値
- 階級の幅dを決める：階級数が7から15程度になるように階級の幅を決めることが多い
- 階級（級下限、級上限、階級値）を決める：階級値=(級下限+級上限)/2
- 度数を求める：度数とは、その級下限以上、その級上限未満の数値をとるデータの数を表す

例：2023年の日本のプロ野球のパ・リーグの打撃成績（100打席以上）の安打数の度数分布表を作る

## Frequency distribution
Procedure to create a frequency distribution
- Obtain the range: (range)=(maximum)-(minimum)
- Determine the classes width d: the class width is often determined so that the number of classes is between 7 and 15.
- Determine the classes (lower class limit, upper class limit, class value): class value = (lower class limit + upper class limit)/2
- Calculate the frequencies: The frequency is the number of data at or above the lower class limit and below the upper class limit.

````python
# 最大値から最小値を引くことで範囲を求める。
# Calculate the range by subtracting the minimum value from the maximum value.
h_range = h_max - h_min
print(f'安打の範囲: {h_range} 本') #Range
````

### 階級の幅dを決める
求めた範囲の中で度数を計算する階級の幅を決める。階級の数を決めることで幅dが求まる。

### Determining the class width d
Determine the width of the class for which the frequencies are calculated within the range obtained. The width d is determined by the number of classes.

````python
# 階級数を10にした場合の幅を求める
# Calculate the width if the number of classes is set to 10.
h_n_ini = 10 #階級数を決める (Determine the number of classes)
d = h_range/h_n_ini #範囲を階級数で割った値が階級の幅d (The range divided by the number of classes is d)
print(f'階級の幅d: {d}') #Class width
````

````python
# 練習 4 階級数として7~15から好きな数を選び、階級の幅を計算してください。
# Exercise 4 Choose any number from 7-15 as the number of classes and calculate the width of the class.
h_n = 7
d = h_range/h_n
print(f'階級の幅d: {d}') #Class width
````

### 階級を決める
階級の幅を決めることで、階級（級下限、級上限、階級値）が決まる。具体的には、最初の階級の級下限を最小値として、幅dを級下限に足すことで級上限が求める。次に級下限と級上限の平均値が階級値として求められる。次の階級の級下限を前の階級の級上限として、この作業を繰り返すことで、全ての階級が求められる。

### Determining the classes
By determining the width of a class, the class (lower class limit, upper class limit and class value) is determined. Specifically, the lower limit of the first class is the minimum value, and the upper limit of the class is obtained by adding the width d to the lower limit of the class. The average of the lower and upper class limits is then determined as the class value. The lower limit of the next class is the upper limit of the previous class, and this process is repeated to obtain all classes.

````python
# 階級数nの場合の階級を計算する。
# Calculate the classes when the number of classes is n.
n = 10 # 階級数 (the number of classes)
min_val = h_min #級下限の初期値に最小値を入れる (Put the minimum value as the initial value of the lower class limit)
max_val = None
c_val = None
print(f'最小値: {h_min}') #Minimum value
print(f'最大値: {h_max}') #Maximum value
for i in range(0,n): #階級の数の回数分繰り返す (Repeat times for the number of classes)
    max_val = min_val + d #級下限に幅dを足す (Add width d to the lower limit of the class)
    c_val = (min_val + max_val) / 2 #級下限と級上限の平均値を求める (Calculate the average of the lower and upper class limits)
    print(f'階級 {i+1}: {c_val:,.1f} ({min_val:,.1f}-{max_val:,.1f})') #Class
    min_val = max_val #次の階級の級下限を前の階級の級上限に更新する (Update the lower class limit of the next class to the upper class limit of the previous class)
````

### 度数を求める
各階級の級下限以上、級上限未満の数値をとるデータの数が度数である。`array`形式のデータから度数分布を求めるためには、関数`np.histogram()`を使う。

---
```Python
{度数},{階級} = np.histogram(a, range=({最小値},{最大値}), bins={階級数})
```
---

### Calculating the frequencies
The frequency is the number of data that are at or above the lower limit of each class and below the upper limit of each class. To obtain the frequency distribution from data in `array` format, use the function `np.histogram()`.

---
```Python
{Frequencies}, {Classes} = np.histogram(a, range=({minimum value},{maximum value}), bins={number of classes})
```
---

````python
# 階級数10の場合の度数を計算する。
# Calculate the frequencies when the number of classes is 10.
hist_h, edge_h = np.histogram(hit, range=(h_min,h_max), bins=h_n_ini)
print('度数') #Frequencies
print(hist_h)
print('階級') #Classes
print(edge_h)
````

## クロス集計表
クロス集計表は異なる2つの属性に着目し、データを分類する方法である。2つの変数に着目して、度数分布表と同様にデータを整理する。度数分布は`Pandas`の`crosstab`関数で作ることができる。

---
```Python
pd.crosstab(df['列名1'], df['列名2'])
```
---

例：ある学校の普通科と国際科の学生の英検2級の取得と留学経験に関するアンケートのクロス集計表を作成する。

## Crosstabulations Tables
A crosstabulation table is a method of classifying data by focusing on two different attributes; it organizes data in the same way as a frequency distribution table, focusing on two variables. Frequency distributions can be created with the `crosstab` function in `Pandas`.

---
```Python
pd.crosstab(df['column name1'], df['column name2'])
```
---

Example: Create a crosstabulation table for a questionnaire about the English proficiency test level 2 and study abroad experience of regular course ('普通') and international course students ('国際') in a certain school.

````python
# 'sample_3-2.csv'を読み込んで、変数'df2'に代入する。
# Read 'sample_3-2.csv' and assign it to variable 'df2'.
df2 = pd.read_csv('sample_3-2.csv', encoding='shift-jis')
df2.head(10)
````

````python
# 全体のクロス集計表を作成する。
# Create a crosstabulation table for the whole students.
pd.crosstab(df2['英検2級の取得'], df2['留学経験'])
# 英検2級の取得: English proficiency test level 2
# 留学経験: Study abroad experience
# 有/無: Yes/No
````

````python
# 普通科の学生のクロス集計表を作成する。
# Create a crosstabulation table for the regular course ('普通') students.
df2_g = df2[df2['クラス']=='普通'] #df2の'クラス'列が'普通'の列を抽出 (Extract columns where the df2 'course enrolment (クラス)' column is 'normal (普通)')
pd.crosstab(df2_g['英検2級の取得'], df2_g['留学経験'])
````
